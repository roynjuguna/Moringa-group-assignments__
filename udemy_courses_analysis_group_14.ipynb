**Problem Statement:**

The objective of this analysis is to conduct a comprehensive evaluation of the course offerings on Udemy to gain insights into their subjects, pricing structures, and popularity. Specifically, we aim to:

1. Identify all the different subjects for which Udemy is offering courses.
2. Determine which subject has the maximum number of courses available.
3. Classify courses into two categories: free and paid.
4. Identify all paid courses and analyze their pricing.
5. Investigate sales data to identify the top-selling and least-selling courses.
6. Filter for graphic design courses priced below $100.
7. Compile a list of courses related to "Python."
8. Extract data on courses published in the year 2015.
9. Analyze the maximum number of subscribers for each level of the courses offered.

This analysis will provide a clearer understanding of the course landscape on Udemy, enabling better decision-making for potential learners and course developers.






```python
# Importing necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Reading the dataset from a CSV file using pandas
data = pd.read_csv(r'udemy_courses_data.csv')
```

```python
# Checking for duplicates
data.duplicated().sum()
```

```python
# Dropping the duplicates
data.drop_duplicates(inplace=True)
```

```python
# Confirming the duplicates are removed
data.duplicated().sum()
```

```python
# Checking for null values
data.isna().sum()
```

```python
# Checking the shape of the dataset
data.shape
```

## What are all the different subjects for which udemy is offering courses?

```python
data.head()
```

```python
# Getting information about our dataset
data.info()
```

```python
# Retrieving the unique values from the 'subject' column of the DataFrame
data['subject'].unique()
```

## Which subject has the maximum number of courses?

```python
# Finding the maximum value in the 'subject' column of the DataFrame
data['subject'].max()
```

## Show all the courses which are free of cost

```python
# Filtering the DataFrame to only include rows where the 'is_paid' column is False
data[data['is_paid'] == False]
```

## Show all the courses which are paid

```python
# Filtering the DataFrame to only include rows where the 'is_paid' column is True
data[data['is_paid'] == True]
```

## Which are the top selling courses?

```python
# Sorting the DataFrame by the 'num_subscribers' column in descending order
data.sort_values('num_subscribers', ascending=False)
```

## Which are the least selling courses?

```python
# Sorting the DataFrame by the 'num_subscribers' column in ascending order
data.sort_values('num_subscribers', ascending=True)
```

## Show all the courses of graphic design where the price is below 100

```python
# Filtering the DataFrame to include only courses in the 'Graphic Design' subject with a price less than 100
data[(data['subject'] == 'Graphic Design') & (data['price'] < '100')]
```

## List all the courses that are related with "Python".

```python
# Filtering the DataFrame to include only courses whose titles contain the word 'Python'
python_courses = data[data['course_title'].str.contains('Python' or 'python')]
python_courses.shape
```

## What are the courses published in the year 2015?

```python
# Displaying the data types of each column in the DataFrame
data.dtypes
```

```python
# Converting the 'published_timestamp' column to datetime format
data['published_timestamp'] = pd.to_datetime(data['published_timestamp'])
```

```python
# Extracting the year from the 'published_timestamp' column and creating a new column 'Year'
data['Year'] = data['published_timestamp'].dt.year
```

```python
# Filtering the DataFrame to include only rows where the 'Year' column is 2015
data[data['Year'] == 2015]
```

## What are the maximum number of subscribers for each level of the courses?

```python
# Retrieving the unique values from the 'level' column of the DataFrame
data['level'].unique()
```

```python
# Grouping the DataFrame by the 'level' column and calculating the maximum number of subscribers for each level
data.groupby('level')['num_subscribers'].max()
```

## Conclusion
- The top selling courses are web development courses
- The least selling courses are Musical instruments, Business finance and graphic design
- The different subjects offered are Musical Instruments', 'Business Finance', 'Graphic Design', 'Web Development'
- 310 courses are free of charge and 3366 courses are paid
- There are no graphic design courses where the price is below 100
- 29 courses have the keyword 'Python'
- 1014 courses were published in 2015
